```python
import matplotlib.pyplot as plt
import numpy as np

# Extract KPI details from the existing DataFrame
kpis = list(zip(kpi_summary['Metric'], kpi_summary['Value']))

# Create dashboard with 2 rows and 5 columns
fig, axes = plt.subplots(2, 5, figsize=(20, 9))

fig.suptitle(
    "INSURANCE ANALYTICS DASHBOARD",
    fontsize=20,
    fontweight='bold'
)

# Loop through each KPI
for ax, (title, value) in zip(axes.flatten(), kpis):

    # Background color
    ax.set_facecolor('#F2EDE5')

    # Format KPI values
    if isinstance(value, (int, np.integer)):
        display_value = f"{value:,}"
    else:
        display_value = f"{value:,.2f}"

    # KPI title
    ax.text(
        0.5, 0.65, title,
        ha='center',
        va='center',
        fontsize=12,
        fontweight='bold',
        color='#75543D'
    )

    # KPI value
    ax.text(
        0.5, 0.35, display_value,
        ha='center',
        va='center',
        fontsize=21,
        fontweight='bold',
        color='#3D3028'
    )

    # Remove axis ticks
    ax.set_xticks([])
    ax.set_yticks([])

    # Card border
    for spine in ax.spines.values():
        spine.set_color('#B89B7A')
        spine.set_linewidth(2)

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()
```
